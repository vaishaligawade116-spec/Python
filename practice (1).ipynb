```python
#grading system
marks=int(input("enter your marks"))

if(marks>90) and (marks<=100):
    print("distinction")
elif(marks>85):
    print("first class")
elif(marks>55):
    print("second class")
elif(marks>35):
    print("pass class")
elif(marks<35) and (marks==0):
    print("fail")
else:
    print("invalid")
```

```python
#rating system
rating=int(input("enter your rating from 1 to 5:"))
if(rating==5):
    print("excellent")
elif(rating==4):
    print("too good")
elif(rating==3):
    print("good")
elif(rating==2):
    print("average")
elif(rating==1):
    print("bad")
else:
    print("invalid")
   
```
