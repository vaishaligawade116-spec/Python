```python
#Bank loan approval → Country → Employment → Salary → Credit score
country=input("are you from india:")
employment=input("are you employed:")
salary=int(input("enter your salary:"))
credit_score=int(input("enter your credit score:"))

if(country=="yes"):
    if(employment=="yes"):
        if(salary>=50000):
            if(credit_score>1000):
                print("bank loan approved")
            else:
                print("your credit score is less")
        else:
            print("your salary is low")
    else:
        print("you are not employed")
else:
    print("you are not from india")
    
```

```python
#College admission → State → Marks → Age → Qualification	
state=input("enter your state:")
marks=int(input("enter your marks:"))
age=int(input("enter your age:"))
qualification=input("enter your education qualification:")

if(state=="belgaum"):
    if(marks>75):
        if(age>=18):
            if(qualification=="12th"):
                print("admission confirmed for 12th pass student")
            else:
                print("admission rejected for 12th fail student")
        else:
            print("admission rejected bcz age is not matching the criteria")
    else:
        print("admission rejected due to marks being less than 75")
else:
    print("admission rejected due to student being from other state")
   
```

```python
#Online shopping → Login → Product availability → Payment → Order confirmation	
login=input("check whether you have logged in:")
product_availability=input("whether prodct availaible:")
payment=input("payment successful:")
order_confirmation=input("enter whether order is confirmed:")

if(login=="yes"):
    if(product_availability=="yes"):
        if(payment=="yes"):
            if(order_confirmation=="yes"):
                print("order confirmed")
            else:
                print("not confirmed")
        else:
            print("payment not successful")
    else:
        print("product not available")
else:
    print("user not login")
```

```python
#Insurance eligibility → Country → Age → Income → Medical/other criteria	
country=input("are you from india:")
age=int(input("enter your age:"))
income=int(input("enter your income:"))
medical=input("do you satisfy any medical criteria:")

if(country=="yes"):
    if(age>=18):
        if(income>=30000):
            if(medical=="yes"):
                print("you are eligible for insurance ")
            else:
                print("you do not satify the medical critera")
        else:
            print("your income is low")
    else:
        print("you are below age criteria")
else:
    print("you are not indian")
    
```

```python
#Employee promotion → Department → Experience → Performance → Salary criteria	
department=input("enter your department:")
experience=int(input("enter your experience:"))
performance=input("is your performance good:")
salary_criteria=int(input("enter your salary:"))

if(department=="IT"):
    if(experience>2):
        if(performance=="yes"):
            if(salary_criteria>=30000):
                print("your promotion is confirmed")
            else:
                print("your salary is less for promotion")
        else:
            print("your performance is not satisfying")
    else:
        print("your experience is not enough for promotion")
else:
    print("you do not belong to IT department")
```
