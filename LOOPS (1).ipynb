### LOOPS

```python
# This cell demonstrates a while loop and how a loop variable controls repetition.
#while loop
i=0
while i<5:
    print(i)
    i=i+1
```

```python
#while loop
i=0
while i<5:
    i=i+1
    print(i)
   
```

```python
#odd numbers from 1 to 15
i=1
while i<16:
    print(i)
    i=i+2
```

```python
#even numbers
i=2
while(i<16):
    print(i)
    i=i+2
```

```python
##10 to 1
i=10
while i>=1:
    print(i)
    i=i-1
```

```python
#continue
i=0
while i<5:
    i=i+1
    if i==3:
        continue
    print(i)
```

```python
#break
i=0
while i<5:
    i=i+1
    if i==3:
        break
    print(i)
```
