```python
# Question 1:rectangle area
def rectangle_area(length, width):
    return length*width
print(rectangle_area(5,3))
    
```

```python
# Question 2:Is Even

def is_even(n):
   return n%2==0
print(is_even(4))

```

```python
# Question 3:Sum of List

def sum_list(numbers):
    return sum(numbers)
print(sum_list([1,2,3]))
```

```python
# Question 4: Reverse String
def reverse_string(s):
    return s[::-1]
print(reverse_string("hello"))
```

```python
# Question 5: Find Maximum
def find_max(numbers):
    return max(numbers)
print(find_max([3, 7, 2, 9, 4]))
```

```python
# Question 6: Square a Number
def square(n):
    return n ** 2
print(square(5))
```

```python
# Question 7: Is Palindrome

def is_palindrome(s):
    return s==s[::-1]
print(is_palindrome("racecar") )
```

```python
# Question 8: Count Vowels
def vowel_count(s):
    count=0
    for char in s:
        if char.lower() in "aeiou":
            count+=1
    return count
print(vowel_count("hello world"))
```

```python
# Question 9: Factorial
def factorial(n):
    res=1
    for i in range(1,n+1):
        res *=i
    return res
print(factorial(5))
    
```

```python
# Question 10: Is Prime
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(7))
```

```python
# Question 11: Celsius to Fahrenheit
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32
print( celsius_to_fahrenheit(0))
```

```python
# Question 12: Count Occurrences
def count_occurrences(lst, item):
    count = 0

    for value in lst:
        if value == item:
            count += 1

    return count

print(count_occurrences([1, 2, 2, 3, 2], 2))
```

```python
# Question 13: Average of List
def average(numbers):
    return sum(numbers) / len(numbers)

print(average([2, 4, 6]))
```

```python
# Question 14: Remove Duplicates
def remove_duplicates(lst):
    result = []

    for item in lst:
        if item not in result:
            result.append(item)

    return result

print(remove_duplicates([1, 2, 2, 3, 1]))
```

```python
# Question 15: FizzBuzz
def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return str(n)

print(fizzbuzz(15))
print(fizzbuzz(9))
print(fizzbuzz(7))
```
